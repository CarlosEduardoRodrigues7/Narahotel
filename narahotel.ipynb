```python
mapa = {
    '4 estrelas':                '4',
    '4 Estrelas':                '4',
    '3*':                        '3',
    'Três Estrelas':             '3',
    '3 estrelas':               '3',
    '5 estrelas':                '5',
    'cinco estrelas':            '5'
}
df['categoria_hotel'] = df['categoria_hotel'].map(mapa).fillna(df['categoria_hotel'])

print('Após mapeamento:')
print(sorted(df['categoria_hotel'].unique()))
```

```python
reg = {
    'capital':                'Capital',
    'CAP':                  'Capital',
    'Serra':                   'Serra',
     'Baixada fluminense':             'Baixada Fluminense'
}
df['regiao'] = df['regiao'].map(reg).fillna(df['regiao'])

print('Após mapeamento:')
print(sorted(df['regiao'].unique()))
```

```python
df['regiao'] = df['regiao'].str.strip().str.title()
df
```

```python
df['num_quartos_total'] = df['num_quartos_total'].astype("Int64")
df


```

```python
df = df.rename(columns={'categoria_hotel': 'categoria_hotel(estrelas)'})
df
```

```python
df['categoria_hotel(estrelas)'] = pd.to_numeric(df['categoria_hotel(estrelas)'], errors="coerce").astype("Int64")
df
```

```python
df_tipo = pd.read_csv('tipos_quarto.csv', sep=';')

print(df_tipo.shape)
display(df_tipo.head())
display(df_tipo.tail())
display(df_tipo.dtypes)
```

```python
df_tipo['descricao'] = df_tipo['descricao'].str.strip().str.title()
df_tipo
```

```python
df_tipo['valor_diaria_base'] = (
df_tipo['valor_diaria_base']
.str.replace('R$', '', regex=False)
.str.replace(' ', '', regex=False)
.str.replace('.', '', regex=False) # remove separador de milhar
.str.replace(',', '.', regex=False)
.astype(float)
)

```

```python
df_tipo.to_csv('tipos_quarto.csv', sep=';', index=False)
```

```python
df_fun = pd.read_csv('funcionarios.csv', sep=';')
df_fun

print(df_fun.shape)
display(df_fun.head())
display(df_fun.tail())
display(df_fun.dtypes)
```

```python
df_fun['cargo'] = df_fun['cargo'].str.strip().str.title()
```

```python
df_fun['cargo'].value_counts()
```

```python
df_fun['departamento'].value_counts()
```

```python
dep = {
    'Gov.':                'Governança',
    'Admin.':                  'Administração',
    'A&B':                   'Alimentos e Bebidas',
     'Baixada fluminense':             'Baixada Fluminense'
}
df_fun['departamento'] = df_fun['departamento'].map(dep).fillna(df_fun['departamento'])

print('Após mapeamento:')
print(sorted(df_fun['departamento'].unique()))
```

```python
df_fun['salario'] = (
df_fun['salario']
.str.replace('.', '', regex=False)
.str.replace('"', '', regex=False)
.str.replace(',', '.', regex=False)
.astype(float)
)
```

```python
df_fun['data_admissao'] = pd.to_datetime(
    df_fun['data_admissao'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)
```

```python
display(df_fun.dtypes)
df_fun.to_csv('funcionarios.csv', sep=';', index=False)
df_fun
```

padronizado departamento tirado abreviações
