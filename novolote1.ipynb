```python
import pandas as pd 
import numpy as np
```

```python
pd.read_csv ('canais_venda.csv', sep=';')
pd.read_csv ('funcionarios_novos.csv', sep=';')
pd.read_csv ('reservas_novas.csv', sep=';')
pd.read_csv ('tipos_quarto_novos.csv', sep=';')
pd.read_csv ('unidades_novos.csv', sep=';')
```

```python
df = pd.read_csv('unidades_novos.csv', sep=';')

print(df.shape)
display(df.head())
display(df.tail())
display(df.dtypes)
```

```python
df['categoria_hotel'].value_counts()
```

```python
df[df.duplicated(subset=['num_quartos_total'], keep=False)]
```

```python
mapa = {
    '5 estrelas':                '5',
    'cinco estrelas':            '5'
}
df['categoria_hotel'] = df['categoria_hotel'].map(mapa).fillna(df['categoria_hotel'])

print('Após mapeamento:')
print(sorted(df['categoria_hotel'].unique()))
```

```python
df['categoria_hotel'].value_counts()
```

```python
df['regiao'].value_counts()
```

```python
df['regiao'] = df['regiao'].str.strip().str.title()
df
```

```python
df['num_quartos_total'] = df['num_quartos_total'].astype("Int64")
df
```

```python
df = df.rename(columns={'categoria_hotel': 'categoria_hotel(estrelas)'})
df
```

```python
df['categoria_hotel(estrelas)'] = pd.to_numeric(df['categoria_hotel(estrelas)'], errors="coerce").astype("Int64")
df
```

```python
df.to_csv('unidades_novos.csv', sep=';', index=False)
df
```

```python
df_tipo = pd.read_csv('tipos_quarto_novos.csv', sep=';')

print(df_tipo.shape)
display(df_tipo.head())
display(df_tipo.tail())
display(df_tipo.dtypes)
```

```python
df_tipo['valor_diaria_base'] = (
df_tipo['valor_diaria_base']
.str.replace('R$', '', regex=False)
.str.replace(' ', '', regex=False)
.str.replace('.', '', regex=False) # remove separador de milhar
.str.replace(',', '.', regex=False)
.astype(float)
)
```

```python
df_tipo.to_csv('tipos_quarto_novos.csv', sep=';', index=False)
```

```python
df_tipo = pd.read_csv('canais_venda.csv', sep=';')

print(df_tipo.shape)
display(df_tipo.head())
display(df_tipo.tail())
display(df_tipo.dtypes)
```

```python
df_tipo['nome_canal'].value_counts()
```

```python
df_tipo['nome_canal'] = df_tipo['nome_canal'].str.strip().str.title()
df_tipo
```

```python
df_tipo['comissao_pct'].value_counts()
```

```python
df_tipo["comissao_pct"] = (
    df_tipo["comissao_pct"]
    .str.replace("%", "", regex=False)
    .str.replace(",", ".", regex=False)
    .astype(float)
)
```
