```python
df = pd.read_csv('funcionarios_novos.csv', sep=';')

print(df.shape)
display(df.head())
display(df.tail())
display(df.dtypes)

```

```python
df['salario'] = (
    df['salario']
    .astype(str)
    .str.replace('"', '', regex=False)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
    .astype(float)
)
df
```

```python
df['data_admissao'] = pd.to_datetime(
    df['data_admissao'],
    format='mixed',
    dayfirst=True
)
df
```

```python
df.to_csv('funcionarios_novos.csv', sep=';', index=False)
df
```
